```python
import numpy as np

# 1. np.array(): Creates an array from standard Python lists
arr_1d = np.array([1, 2, 3, 4, 5])
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])

# 2. np.zeros(): Creates an array filled with zeros of a specified shape
zeros_arr = np.zeros((2, 3))  # 2 rows, 3 columns

# 3. np.ones(): Creates an array filled with ones
ones_arr = np.ones((3, 3), dtype=int)

# 4. np.full(): Creates an array filled with a specified constant value
full_arr = np.full((2, 2), 7)  # fills a 2x2 matrix with 7s

# 5. np.arange(): Generates sequences of numbers with a given step
range_arr = np.arange(0, 10, 2)  # Output: [0, 2, 4, 6, 8]

# 6. np.linspace(): Generates specified number of evenly spaced values over an interval
linspace_arr = np.linspace(0, 1, 5)  # Output: [0.  , 0.25, 0.5 , 0.75, 1.  ]

# 7. np.eye(): Creates a 2D Identity Matrix
identity_matrix = np.eye(3)

print("1D Array:\n", arr_1d)
print("Zeros Array:\n", zeros_arr)
print("Linspace Array:\n", linspace_arr)
```

```python
# Create a sample 2D array for testing properties
sample_arr = np.array([[10, 20, 30], [40, 50, 60]])

# 8. .shape: Returns tuple representing array dimensions
print("Shape:", sample_arr.shape)  # Output: (2, 3)

# 9. .ndim: Returns number of array dimensions (axes)
print("Dimensions:", sample_arr.ndim)  # Output: 2

# 10. .dtype: Returns the data type of the elements
print("Data Type:", sample_arr.dtype)  # Output: int64 (or int32 depending on OS)

# 11. .size: Returns total number of elements in the array
print("Total Elements:", sample_arr.size)  # Output: 6

# 12. .astype(): Casts an array to a different data type
float_arr = sample_arr.astype(np.float64)
print("Casted Array Type:", float_arr.dtype)
```

```python
base_arr = np.arange(12)  # [0, 1, ..., 11]

# 13. np.reshape(): Changes the layout of an array without changing its data
reshaped = np.reshape(base_arr, (3, 4))

# 14. .flatten(): Collapses a multi-dimensional array into a 1D copy
flat_arr = reshaped.flatten()

# 15. np.transpose() / .T: Swaps the axes of an array (transpose)
transposed = reshaped.T

# 16. np.concatenate(): Joins a sequence of arrays along an existing axis
arr_a = np.array([[1, 2], [3, 4]])
arr_b = np.array([[5, 6], [7, 8]])
concat_row = np.concatenate((arr_a, arr_b), axis=0)  # Stack vertically
concat_col = np.concatenate((arr_a, arr_b), axis=1)  # Stack horizontally

# 17. np.vstack() & np.hstack(): Shortcut functions for stacking vertical / horizontal
v_stack = np.vstack((arr_a, arr_b))
h_stack = np.hstack((arr_a, arr_b))

# 18. np.split(): Splits an array into multiple sub-arrays
split_arr = np.split(base_arr, 3)  # Splits into 3 equal arrays

print("Reshaped (3x4):\n", reshaped)
print("Transposed (4x3):\n", transposed)
```

```python
calc_arr = np.array([[1, 2, 3], [4, 5, 6]])

# 19. np.sum(): Calculates the sum of elements (can evaluate over specific axes)
print("Total Sum:", np.sum(calc_arr))
print("Sum over columns (axis=0):", np.sum(calc_arr, axis=0))  # Output: [5, 7, 9]

# 20. np.mean(): Computes the arithmetic mean
print("Mean:", np.mean(calc_arr))

# 21. np.std() & np.var(): Compute Standard Deviation and Variance
print("Standard Deviation:", np.std(calc_arr))

# 22. np.min() & np.max(): Find absolute minimum and maximum values
print("Max Value:", np.max(calc_arr))

# 23. np.cumsum(): Computes cumulative sum of elements
print("Cumulative Sum:", np.cumsum(calc_arr))
```

```python
math_arr = np.array([0, np.pi/2, np.pi])

# 24. Trigonometric functions: np.sin(), np.cos(), np.tan()
print("Sine values:", np.sin(math_arr))

# 25. Exponential and Logarithmic functions
nums = np.array([1, 2, 3])
print("Exponential (e^x):", np.exp(nums))
print("Natural Log (ln):", np.log(nums))

# 26. np.sqrt(): Computes element-wise square root
print("Square Root:", np.sqrt(np.array([4, 9, 16])))

# 27. np.abs(): Computes absolute values
print("Absolute Values:", np.abs(np.array([-1, -2, 3])))
```

```python
matrix_x = np.array([[1, 2], [3, 4]])
matrix_y = np.array([[5, 6], [7, 8]])

# 28. np.dot(): Dot product of two arrays
dot_product = np.dot(matrix_x, matrix_y)

# 29. np.matmul() or @ operator: Matrix multiplication
matrix_mult = matrix_x @ matrix_y

# 30. np.linalg.inv(): Computes the inverse of a square matrix
inv_matrix = np.linalg.inv(matrix_x)

# 31. np.linalg.det(): Computes the determinant of a matrix
det_value = np.linalg.det(matrix_x)

print("Matrix Multiplication (@):\n", matrix_mult)
print("Matrix Determinant:", det_value)
```
