```python
# Select and store specific rows (1, 0, 6, 8, 4) from DataFrame 'ab' into 'dr' using .loc
dr = ab.loc[[1,0,6,8,4]]
# Print the selected rows
print(dr)
# Display the selected rows
display(dr)
```

```python
# Create a dummy column 'dummy_column' with a range of values
ab['dummy_column'] = range(len(ab))

# Insert a new column 'new_column_inserted' at index 1 (second position)
# The values are twice the 'dummy_column' values
ab.insert(1, 'new_column_inserted', ab['dummy_column'] * 2)
# Display the head of the modified DataFrame 'ab'
display(ab.head())
```

```python
# Use assign to create new columns 'amount_category' and 'time_category'
# 'amount_category' categorizes 'amount' into 'Low', 'Medium', 'High' based on bins
# 'time_category' categorizes 'time_gap_minutes' into 'Short', 'Medium', 'Long' based on bins

ab_assigned = ab.assign(
    amount_category=lambda x: pd.cut(x['amount'], bins=[0, 10000, 20000, 50000], labels=['Low', 'Medium', 'High']),
    time_category=lambda x: pd.cut(x['time_gap_minutes'], bins=[0, 500, 1000, 1500], labels=['Short', 'Medium', 'Long'])
)
# Display the head of the new DataFrame 'ab_assigned' with the new columns
display(ab_assigned.head())
```
