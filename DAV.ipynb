```python
# Program 1(a)
# Create 1D, 2D and 3D NumPy arrays
import numpy as np
a1 = np.array([1, 2, 3])
a2 = np.array([[1, 2], [3, 4]])
a3 = np.array([[[1, 2]], [[3, 4]]])
print(a1, a2, a3)
```

```python
# Program 1(b)
# Array creation using zeros, ones and empty
import numpy as np
print(np.zeros((2,2)), np.ones((2,2)), np.empty((2,2)))
```

```python
# Program 2
# NumPy array attributes (ndim, shape, size, dtype)
import numpy as np
a = np.array([[1, 2, 3], [4, 5, 6]])
print(a.ndim, a.shape, a.size, a.dtype)
```

```python
# Program 3
# Slicing and indexing NumPy arrays
import numpy as np
a = np.array([[1, 2, 3], [4, 5, 6]])
print(a[0, 1], a[:, 1:3])
```

```python
# Program 4
# Matrix Operations: Addition, Multiplication, Transpose
import numpy as np
m1 = np.array([[1, 2], [3, 4]])
m2 = np.array([[5, 6], [7, 8]])
print(m1 + m2, np.dot(m1, m2), m1.T)
```

```python
# Q2(a)
# Structure, datatypes, and summary stats of student dataset
import pandas as pd
data = {'Roll_No': [1, 2], 'Maths': [85, 90], 'Python': [88, 92], 'DBMS': [82, 89]}
df = pd.DataFrame(data)
print(df.info())
print(df.describe())
```

```python
# Q3(a)
# Find null values in student dataset and handle them
import pandas as pd
import numpy as np
data = {'Name': ['Amit', 'Bina'], 'Maths': [85, np.nan]}
df = pd.DataFrame(data)
print(df.isnull().sum())
print(df.fillna(df['Maths'].mean()))
```

```python
# Q4(a)
# Filter students with Attendance > 75%
import pandas as pd
data = {'Name': ['Amit', 'Bina', 'Chetan'], 'Attendance': [80, 95, 60]}
df = pd.DataFrame(data)
print(df[df['Attendance'] > 75])
```

```python
# Q5(a)
# Sort students by Python marks descending
import pandas as pd
data = {'Name': ['Amit', 'Bina', 'Chetan'], 'Python': [88, 92, 40]}
df = pd.DataFrame(data)
print(df.sort_values(by='Python', ascending=False))
```

```python
# Q6(a)
# Average marks of Maths per department
import pandas as pd
data = {'Department': ['CS', 'IT', 'CS'], 'Maths': [85, 90, 45]}
df = pd.DataFrame(data)
print(df.groupby('Department')['Maths'].mean())
```

```python
# Q7(a)
# Count of male and female students in each department
import pandas as pd
data = {'Department': ['CS', 'IT', 'CS'], 'Gender': ['M', 'F', 'M']}
df = pd.DataFrame(data)
print(pd.crosstab(df['Department'], df['Gender']))
```

```python
# Q8(a)
# Create a new column Total_Marks (Maths + Python + DBMS)
import pandas as pd
data = {'Maths': [85, 90], 'Python': [88, 92], 'DBMS': [82, 89]}
df = pd.DataFrame(data)
df['Total_Marks'] = df['Maths'] + df['Python'] + df['DBMS']
print(df)
```

```python
# Q14(a)
# Outlier detection in Study_Hours using Z-score
import pandas as pd
import numpy as np
from scipy import stats
data = {'Study_Hours': [6, 8, 2, 5, 25]}
df = pd.DataFrame(data)
z_scores = np.abs(stats.zscore(df['Study_Hours']))
print("Outliers:", df[z_scores > 1.5])
```

```python
# Q17(a)
# Pivot table: Mean Python marks by Department and Gender
import pandas as pd
data = {'Department': ['CS', 'CS', 'IT'], 'Gender': ['M', 'F', 'F'], 'Python': [88, 92, 80]}
df = pd.DataFrame(data)
print(df.pivot_table(index='Department', columns='Gender', values='Python', aggfunc='mean'))
```
