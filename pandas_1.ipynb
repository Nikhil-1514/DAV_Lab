```python
df_result = pd.read_csv('students.csv')
df_result['Result'] = np.where(df_result['Marks'] >= 40, 'Pass', 'Fail')
df_result.to_csv('result.csv', index=False)
print("DataFrame with 'Result' column saved to result.csv:")
display(df_result)
```

```python
df_grade = pd.read_csv('students.csv')

def assign_grade(marks):
    if marks >= 90:
        return 'A'
    elif marks >= 75:
        return 'B'
    elif marks >= 50:
        return 'C'
    else:
        return 'D'

df_grade['Grade'] = df_grade['Marks'].apply(assign_grade)

print("DataFrame with 'Grade' column:")
display(df_grade)

print("\nNumber of students in each grade:")
display(df_grade['Grade'].value_counts().sort_index())

print("\nDepartment-wise average marks:")
display(df_grade.groupby('Department')['Marks'].mean())

overall_average_marks = df_grade['Marks'].mean()
students_above_overall_average = df_grade[df_grade['Marks'] > overall_average_marks]
print(f"\nStudents whose marks are above the overall average ({overall_average_marks:.2f}):")
display(students_above_overall_average)
```
